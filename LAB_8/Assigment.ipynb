Q-1.WAAP to draw a scatter plot using kaggle

```python
import pandas as pd 
import matplotlib.pyplot as plt
data=pd.read_csv("E:\STUDY METERIAL\CSV FILE\Paris 2024 Olympics_Nations Medals.csv")
print(data.head(10))
```

```python
plt.scatter(data["Silver"],data["Gold"])
plt.title("Scatter plot")
plt.xlabel("silver")
plt.ylabel("Gold")
plt.show()
```

Q-2.Draw a line chart 

```python
plt.plot(data["Gold"])
plt.plot(data["Silver"])

plt.title("Line Chart")
plt.xlabel("Gold")
plt.ylabel("Silver")
plt.show()
```

Q-3.Draw Bar chart

```python

plt.bar(data["Bronze"],data["Gold"])
plt.title("Bar chart")
plt.xlabel("Bronze")
plt.ylabel("Gold")
plt.show()
```
