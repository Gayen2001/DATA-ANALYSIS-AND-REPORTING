3.Handling Duplicate values in pandas

**duplicates values alweys find uniqe column or primary key

```python
#Identify duplicate value
print(dataFrame.duplicated()) #its provide duplicate value present or not in a row 
```

```python
#Identify duplicate value in a particular column

print(dataFrame["NOC"].duplicated()) # given data frame NOC is a primary key
#true--> present
#fales-->not present
```

```python
# calculate total number of duplicate value in a column
print(dataFrame["NOC"].duplicated().sum())
```

```python
#drop duplicate value 
print(dataFrame.drop_duplicates("NOC"))
```

4.Working with Missing data in pandas

```python
#find missing value or null value 
dataFrame.isnull()
```

```python
#total number of missing value as column wish 
dataFrame.isnull().sum()
```

```python
# calculate total number of missing value in our data set
dataFrame.isnull().sum().sum()
```

```python
# calculate parcentage of missing value present in each column
dataFrame.isnull().sum()/df.shape[0]*100 #[0] are use for calculate number of row only
```

```python
# total percentage of missing value present in data set
dataFrame.isnull().sum().sum()/(df.shape[0]* df.shape[1])*100
```

```python
#drop missing value
dataFrame.dropna()
```

```python
#replace missing value with any value in full dataframe
import numpy as np

print(dataFrame.replace(np.nan,"hello"))
```

```python
# calulate mean of the total column
dataFrame["Total"].mean()
```

```python
#replace missing value in a particular column  with mean value
dataFrame["Total"]=dataFrame["Total"].replace(np.nan,11.64)
print(dataFrame)

# calulate mean
dataFrame["Total"].mean() 
# does not affect the total mean
```

```python
# fill the missing value in backword method
print(dataFrame.fillna(method="bfill"))
```

```python
# fill the missing value in forword method
print(dataFrame.fillna(method="ffill"))
```
