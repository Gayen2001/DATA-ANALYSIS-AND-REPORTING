data structure in pandas is series(only one column ) and data frame(multiple column)

1.creating DataFrame in pandas

```python
# import pandas library
import pandas as pd
# input data
data={
    "Name":["partha","peter","biswajit"],
    "age":[24,43,22],
    "salary":[20000,40000,21000]
}
# create the above data as data frame 
df = pd.DataFrame(data)

# print the data frame
print(df)
```
